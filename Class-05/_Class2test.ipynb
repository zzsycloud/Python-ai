# 类-热身

编写一个Rectangle类，用于表示一个矩形，包含计算矩形面积和周长的方法。

```python
class Rectangle:
    # 构造函数
    # 对当前对象的实例的初始化
    def __init__(self, length, width):
        """
        初始化矩形的长和宽
        :param length: 矩形的长
        :param width: 矩形的宽
        """
        self.length = length
        self.width = width
 
    def area(self):
        """
        计算矩形的面积
        :return: 矩形的面积
        """
        return self.length * self.width
 
    def perimeter(self):
        """
        计算矩形的周长
        :return: 矩形的周长
        """
        return 2 * (self.length + self.width)
 
# 测试代码
rect = Rectangle(5, 3)
print(f"矩形的面积为: {rect.area()}")
print(f"矩形的周长为: {rect.perimeter()}")
```

一、定义一个学生Student类。有下面的类属性：      

1 姓名 nam           e
2 年龄 a       ge
3 成绩 score（语文，数学，       [) [每课成绩的类型为   数]


类方法：
1 获取学生的姓名：get_name() 返回类型:str
2 获取学生的年龄：get_age() 返回类型:int
3 返回3门科目中最高的分数。get_course() 返回类型:int

```python
class Student():
    # 构造函数
    # 对当前对象的实例的初始化
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

    #   isinstance函数判断一个对象是否是一个已知的类型，类似type
    def get_name(self):
        if isinstance(self.name, str):
            return self.name

    def get_age(self):
        if isinstance(self.age, int):
            return self.age

    def get_course(self):
        a = max(self.score)
        if isinstance(a, int):
            return a

zm = Student('zhangming', 20, [69, 88, 100])
print(zm.get_name())
print(zm.get_age())
print(zm.get_course())
```

二、创建一个类Circle，具有radius实例属性

编写打印方法，返回属性值
编写实例方法，输出圆的面积
编写析构方法，输出"圆的半径不能为负数"

```python
import math

class Circle:
    def __init__(self,radius):
        self.radius = radius
 
    def show(self):
        return math.pi*self.radius**2
 
    def __del__(self):
        if self.radius<0:
            print(f"圆的半径不能为负数")
 
c1 = Circle(2)
s1 = c1.show()
print(f"{s1:.2f}")
```

四、继承
定义名为Fruit的类
编写构造方法，接收1个外部参数color
编写打印方法，返回属性值
编写Fruit的子类Apple，定义方法setColor，给color赋值
实例化子类对象，调用setColor方法个功能

```python
class Fruit:
    def __init__(self,color):
        self.color = color
 
    def __str__(self):
        return self.color
 
class Apple(Fruit):
    def setColor(self,color1):
        self.color = color1

a1 = Apple("red")
print(f"{a1}")
a1.setColor("blue")
print(f"修改后:{a1}")
```

五、封装与私有属性

定义一个BankAccount类，包含私有属性__balance，以及方法deposit和withdraw。

__balance表示当前余额，当前余额不应该轻易让人发现 [私有属性]
deposit方法表示存款，每次存完数额之后返回当前余额
withdraw方法表示取款，每次取完钱之后返回当前余额

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"存款成功，当前余额：{self.__balance}")
        else:
            print("存款金额必须大于0")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"取款成功，当前余额：{self.__balance}")
        else:
            print("取款金额无效或余额不足")

    def get_balance(self):
        return self.__balance

# 实例化并操作
my_account = BankAccount(1000)
my_account.deposit(500)
my_account.withdraw(200)
print(f"当前余额：{my_account.get_balance()}")
```
